### Q5
Given:
```python
A = {2, 4, 6, 8}
B = {4, 8, 12}
```
Find Union, Intersection, A-B

````python
A = {2, 4, 6, 8}
B = {4, 8, 12}

print("Union:", A | B)
print("Intersection:", A & B)
print("A-B:", A - B)
````
